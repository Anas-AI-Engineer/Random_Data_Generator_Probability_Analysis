```python
import numpy as np

rng = np.random.default_rng()

# 1 Random Data Generator

data = {
    
     "Normal": rng.normal(50, 10, 10000), # rng.normal = 50 average no lo and 10 std no lo aur 10000 no generate karo
     "Uniform": rng.normal(0, 100, 10000), # rng.uniform matlab 0 atsrting limit and 100 ending limit 1000 values generate karo
     "Binomial": rng.binomial(10, 0.5, 1000), # rng.binomial matlab 10 bar cions uchalo and 0.5% change ha heads any ky and than 10000 
                                              #ye function repeat karo
     "Poisson": rng.poisson(5, 10000), # rng.possion 5 average per minute 10000 minutes generate karo
}

# 2 statistical Analysis

def statistical_analysis(values): # function banaya statisticalaalysis name rakha

    return{
        "Mean": np.mean(values),
        "Median": np.median(values),
        "Varience": np.var(values),
        "Standard Derivation": np.std(values),
        "Minimium": np.min(values),
        "Maximium":  np.max(values),
        "25 percentile": np.percentile(values, 25),
        "75 percentile": np.percentile(values, 75)
    }

print("===STATISTICAL ANALYSIS====")

for name, values in data.items(): # for loop lagayi name = distribution ka name batyi ga jesy normal values = 12.8 data.items key and values
                                  # dono dekhy ga
    
    print(f"{name} Distribution") # disribution ka name print krwaya

    results = statistical_analysis(values) # variable banya recall krwaya statistics ko

    for metric, values in results.items(): # for loop lagayi metric = mean values = 23.9 results.items = items ki key and values dono dekhaye ga 
        print(f"{metric}, {values:.4f}") # recalls varaibles
        
#Experimental vs theoritcal probability

rolls = rng.integers(1, 7, size=10000)

succeess = np.sum(rolls == 6)

experimental_probability = succeess / len(rolls)
theoretical_probability = 1 / 6

print("Total Rolls:", len(rolls))
print("Total Sixes:", succeess)
print("Experimental probability:", experimental_probability)
print("Theoretical probabity:", theoretical_probability)

print(
    "absolute error:",
    abs(experimental_probability - theoretical_probability)
)

#5 cooleration analysis

cooleration = np.corrcoef( # varaible banaya 
    data["Normal"], # data ko recall krwaya and normal ko
    data["Uniform"], # data ko recall krwaya and uniform ko
)[0,1] # 0 matlab phela columns and 1 matlab dusri row

print("====COOLERATION ANALYSIS====")
print("Normal Vs Uniform", cooleration) # cooleration ko recall krway

#outlier iqr
print("="*40)
values = data["Normal"]# values ky name ky variable banaya # data and normal ko recall krwaya

q1 = np.percentile(values, 25) #q1 ky name sy variable banaya np.percentile percetage maloom karey ga values ko recall krwaya 2 maloom karo 
q3 = np.percentile(values, 75) #q3 ky name sy variable banaya np.percentile percetage maloom karey ga values ko recall krwaya 2 maloom karo

iqr = q3 - q1 #iqr = variable aur kaha ky q3 ma sy - krdu q1 ko

lower_limit = q1 - 1.5 * iqr # lower_limit = variable banay and q1 ma sy - krdu 1.5 and *iqr sy
upper_limit = q3 + 1.5 * iqr # upper_limit = variable banay and q3 ma sy + krdu 1.5 and *iqr sy

outlier = values[ # outlier = variablr banaya and and values ko recall krwaya
    (values < lower_limit) | # values zada ho lowest_limit sy
    (values > upper_limit) # values kam ho upper_limit sy
    ]
print("Outlier pi")
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Outlier:", outlier)
print("="*40)
# 7. MONTE CARLO SIMULATION TO ESTIMATE PI

pionts = rng.uniform(0, 1, size=(10000, 2))

inside_circle = (
    pionts[:,0] ** 2 + pionts[:,1] **2
)

estimate_pi = 4 * np.mean(inside_circle)

print("===MONTE CARLO SIMULATION TO ESTIMATE PI======")
print("Estimate pi:", estimate_pi)
print("Actual pi:", np.pi)
print("Absolute error:", abs(estimate_pi-np.pi))

# frequently analysis
print("====FREQUENTLY ANALYSIS====")

for name, values in data .items():

    unique_values, counts = np.unique(
        values,
        return_counts=True
    )

    print(f"{name}Distribution")
    print("Unique values:", len(unique_values))
    print("Most frequency values:")

    top_incidies = np.argsort(counts)[-5:] [::-1]

    for index in top_incidies:
        print(
            f"values: {unique_values[index]}, "
            f"Frequently: {counts[index]}"
        )

# dice rolls

dice_rolls = rng.integers(1, 7, size=10000)

comulative_sixes = np.cumsum(dice_rolls == 6)
rolls_numbers = np.arange(1 len(dice_rolls) +1 )

runing_probability = comulative_sixes / rolls_numbers

for n [0, 10, 100, 10000, 100000]:
    print(
        f"after" n {rolls}
    )
        
```
